```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, MinMaxScaler
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import confusion_matrix, accuracy_score
```

```python
df = pd.read_csv('data/water_potability.csv')
df
```

```python
X = df.drop("Potability", axis=1)
y = df["Potability"]
```

```python
X = X.fillna(X.median())
X
```

```python
scaler = MinMaxScaler()
X_scaled = scaler.fit_transform(X)
X_scaled
```

```python
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42, stratify=y)
```

```python
from sklearn.metrics import accuracy_score

k_values = range(1, 30)
accuracy_list = []

for k in k_values:
    knn = KNeighborsClassifier(n_neighbors=k)
    knn.fit(X_train, y_train)
    preds = knn.predict(X_test)
    acc = accuracy_score(y_test, preds)
    accuracy_list.append(acc)
    print(f"K = {k} --> Accuracy = {acc}")

best_k = k_values[np.argmax(accuracy_list)]
best_acc = max(accuracy_list)

print("\n==============================")
print("Best K:", best_k)
print("Best Accuracy:", best_acc)
print("==============================")

# train final KNN model with best K
best_knn = KNeighborsClassifier(n_neighbors=best_k)
best_knn.fit(X_train, y_train)
final_preds = best_knn.predict(X_test)

print("\nConfusion Matrix for Best KNN:")
print(confusion_matrix(y_test, final_preds))
```

```python
kernels = ["linear", "rbf", "poly", "sigmoid"]

print("SVM RESULTS\n")

for k in kernels:
    model = SVC(kernel=k)
    model.fit(X_train, y_train)
    preds = model.predict(X_test)

    acc = accuracy_score(y_test, preds)
    cm = confusion_matrix(y_test, preds)

    print(f"\nKernel: {k.upper()}")
    print("Accuracy:", acc)
    print("Confusion Matrix:\n", cm)
```

```python
from sklearn.metrics import classification_report

rbf_model = SVC(kernel='rbf')
rbf_model.fit(X_train, y_train)

# Predictions
y_pred_rbf = rbf_model.predict(X_test)
# Classification Report
print("Classification Report (RBF SVM):")
print(classification_report(y_test, y_pred_rbf))
```
